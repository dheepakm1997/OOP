```python
class Atm:
    # Static Variable
    __customer_id = 0

    def __init__(self):
        self.pin = ""
        self.__balance = 0  # private variable
        self.customer_id = Atm.__customer_id
        Atm.__customer_id += 1

    # Getter method
    def get_balance(self):
        return self.__balance

    @staticmethod
    def get_customer_id():
        return Atm.__customer_id

    # setter method
    def set_balance(self, new_balance):
        if type(new_balance) == int:
            self.__balance = new_balance
        else:
            print("Wrong balance!!")

    def menu(self):
        user_input = input("""
1) Press 1 to create pin
2) Press 2 to change pin
3) Press 3 to check balance
4) Press 4 to withdraw
5) Anything else to exit
""")

        if user_input == "1":
            self.create_pin()
        elif user_input == "2":
            self.change_pin()
        elif user_input == "3":
            self.check_balance()
        elif user_input == "4":
            self.withdraw()
        else:
            print("Thank you for using the ATM.")

    def create_pin(self):
        user_pin = input("Enter your pin: ")
        self.pin = user_pin

        user_balance = int(input("Enter your balance: "))
        self.set_balance(user_balance)

        print("Pin created successfully")
        # self.menu()

    def change_pin(self):
        old_pin = input("Enter your old pin: ")

        if old_pin == self.pin:
            new_pin = input("Enter your new pin: ")
            self.pin = new_pin
            print("Pin changed successfully")
        else:
            print("You have entered an invalid pin")

        # self.menu()

    def check_balance(self):
        user_pin = input("Enter your pin: ")

        if user_pin == self.pin:
            print(f"Your balance is {self.__balance}")
        else:
            print("Your pin is incorrect, please try again")

        # self.menu()

    def withdraw(self):
        user_pin = input("Enter your pin: ")

        if user_pin == self.pin:
            amount = int(input("Enter the amount: "))

            if amount <= self.get_balance():
                self.set_balance(self.get_balance() - amount)
                print(f"Your amount has been withdrawn and the balance is {self.get_balance()}")
            else:
                print("Insufficient balance")
        else:
            print("Your pin is incorrect")

        # self.menu()


atm = Atm()
atm.menu()
```
