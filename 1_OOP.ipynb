```python



class Atm:
    def __init__(self):
        self.pin=""
        self.balance= 0

    def menu(self):
        user_input=input(
            """
            1)Press 1 to create pin
            2)Press 2 to change pin
            3)Press 3 to check balance
            4)Press 4 to withdraw
            5)Anything to exit
            """
        )

        if user_input=="1":
            self.create_pin()
        elif user_input=="2":
            self.change_pin()
        elif user_input=="3":
            self.check_balance()
        elif user_input=="4":
            self.withdraw()
        else:
            exit()
    def create_pin(self):
        user_pin=input("Enter your pin: ")
        self.pin=user_pin

        user_balance =input("Enter your balance: ")
        self.balance=int(user_balance)
        
        print("Pin created successfully ")
        self.menu()
    def change_pin(self):
        old_pin=input("Enter your old pin: ")
        if old_pin == self.pin:
            new_pin=input("Enter your new pin: ")
            self.pin = new_pin
            print("Pin changed successfully")
            self.menu()
        else:
            print("YOu have entered invalid pin")
            self.menu

    def check_balance(self):
        user_pin = input("Enter your pin: ")
        if user_pin == self.pin:
            print(f"Your balance is {self.balance}")
        else:
            print("Your pin is incorrect, please try again")
            self.menu()

    def withdraw(self):
        user_pin = input("Enter your pin: ")
        if user_pin == self.pin:
            amount = int(input("Enter the amount"))
        if amount>=self.pin:
            self.balance=self.balance - amount
            print(f"Your amount has been withddran and the balance is {self.balance}")
        else:
            print("your pin is wrong")

```
